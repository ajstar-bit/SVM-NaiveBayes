# Support Vector Machines & Naive Bayes

## Theoretical Questions

1. What is a Support Vector Machine (SVM)?
- An SVM is a supervised machine learning algorithm used for classification and regression. Its objective is to find the optimal hyperplane that maximally separates data points of different classes in the feature space, thereby maximizing the margin.

2. What is the difference between Hard Margin and Soft Margin SVM?
- Hard Margin SVM: Requires the data to be perfectly linearly separable. It finds a hyperplane that separates the classes with the largest margin, allowing absolutely no misclassification or points within the margin. It is highly sensitive to outliers.
- Soft Margin SVM: Used for non-linearly separable data. It introduces a slack variable ($\xi$) and a penalty parameter ($\mathbf{C}$) to allow some data points to be misclassified or fall within the margin, balancing the trade-off between maximizing the margin and minimizing training errors.

3. What is the mathematical intuition behind SVM?
- The intuition is to formulate the classification problem as a constrained optimization problem:$$\text{Maximize } \frac{2}{\lVert \mathbf{w} \rVert}$$$$\text{Subject to } y_i(\mathbf{w} \cdot \mathbf{x}_i + b) \geq 1$$This aims to maximize the margin $\left( \frac{2}{\lVert \mathbf{w} \rVert} \right)$ while ensuring every data point $\mathbf{x}_i$ is classified correctly (i.e., its distance from the hyperplane is at least 1).

4. What is the role of Lagrange Multipliers in SVM?
- Lagrange Multipliers ($\alpha_i$) are a mathematical technique used to solve the constrained optimization problem of SVM. They allow the problem to be transformed from the primary form into its easier-to-solve dual form, which is crucial because the dual form only depends on the dot product of the feature vectors, enabling the use of the Kernel Trick.

5. What are Support Vectors in SVM?
- Support Vectors are the data points that lie closest to the decision boundary (hyperplane). They are the critical training examples that define the width of the margin and the position and orientation of the optimal hyperplane. If these vectors were removed, the hyperplane would change.

6. What is a Support Vector Classifier (SVC)?
- SVC is the application of the SVM algorithm specifically for classification tasks (e.g., separating data into categories like 'spam' or 'ham').

7. What is a Support Vector Regressor (SVR)?
- SVR is the application of the SVM algorithm specifically for regression tasks. Instead of finding a boundary that separates classes, SVR finds a hyperplane that best fits the data points within a defined margin of tolerance ($\epsilon$-insensitive tube), minimizing the error outside this margin.

8. What is the Kernel Trick in SVM?
- The Kernel Trick is a technique that allows SVMs to create complex, non-linear decision boundaries without explicitly mapping the data into a high-dimensional feature space. It uses a kernel function $K(\mathbf{x}_i, \mathbf{x}_j)$ to calculate the dot product between data points in the high-dimensional space ($\phi(\mathbf{x}_i) \cdot \phi(\mathbf{x}_j)$) more efficiently in the original space.

9. Compare Linear Kernel, Polynomial Kernel, and RBF Kernel.
- | Kernel Type | Formula/Function | Use Case |
| :--- | :--- | :--- |
| Linear | $K(\mathbf{x}_i, \mathbf{x}_j) = \mathbf{x}_i \cdot \mathbf{x}_j$ | Used when data is linearly separable. Fastest and simplest. |
| Polynomial| $K(\mathbf{x}_i, \mathbf{x}_j) = (\gamma (\mathbf{x}_i \cdot \mathbf{x}_j) + r)^d$| Effective for non-linear data where a curved boundary is needed. |
| RBF (Radial Basis Function)| $K(\mathbf{x}_i, \mathbf{x}_j) = \exp(-\gamma \lVert \mathbf{x}_i - \mathbf{x}_j \rVert^2)$| Excellent for handling complex, non-linear patterns by mapping data to an infinite-dimensional space. |

10. What is the effect of the C parameter in SVM?
- The $\mathbf{C}$ parameter is the regularization parameter in Soft Margin SVM. It controls the penalty for misclassification errors:Large C: High penalty, leads to a smaller margin and fewer misclassifications on training data (low bias, high risk of overfitting).
- Small C: Low penalty, leads to a larger margin and allows more misclassifications (high bias, lower risk of overfitting).

11. What is the role of the Gamma parameter in RBF Kernel SVM?
- The Gamma ($\gamma$) parameter defines the influence of a single training example on the decision boundary when using the RBF kernel.Large $\gamma$: Influence only reaches nearby points, leading to a complex, tight, and often overfitting boundary.Small $\gamma$: Influence reaches far away, leading to a smoother, simpler boundary (high bias).

12. What is the Naïve Bayes classifier, and why is it called "Naive"?
- Naive Bayes is a simple probabilistic classifier based on Bayes' Theorem. It's called "Naive" because it makes the fundamental, often unrealistic, assumption that all features are conditionally independent of each other given the class label.

13. What is Bayes' Theorem?
- Bayes' Theorem is a mathematical formula used to determine the conditional probability of an event, calculated using prior knowledge. It is expressed as:$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$In classification, this is:$$P(\text{Class}|\text{Features}) = \frac{P(\text{Features}|\text{Class}) \cdot P(\text{Class})}{P(\text{Features})}$$

14. Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes.

- Gaussian Naïve Bayes: Used when features are continuous and assumed to follow a Gaussian (Normal) distribution.

- Multinomial Naïve Bayes: Used when features represent discrete counts (e.g., word frequency in a document).

- Bernoulli Naïve Bayes: Used when features are binary or Boolean (e.g., the presence or absence of a word in a document).

15. When should you use Gaussian Naïve Bayes over other variants?
- Use Gaussian Naïve Bayes when your input features are continuous numerical values (like height, salary, or measurement data) and you can reasonably assume that these values for each class are distributed normally.

16. What are the key assumptions made by Naïve Bayes?
- The primary assumption is the conditional independence of features. This means that the influence of any single feature value on a given class is independent of the values of the other features.

17. What are the advantages and disadvantages of Naïve Bayes?

- Advantages: It is simple, extremely fast to train and predict, and performs well even with high-dimensional data, often requiring little training data.

- Disadvantages: It suffers from the Zero-Frequency problem, and its "naïve" independence assumption often results in poor probability estimates (though classification is typically robust).

18. Why is Naïve Bayes a good choice for text classification? It is effective for text classification because:

- Text data (like the Bag-of-Words model) is easily modeled by Multinomial or Bernoulli distributions.

- The model handles a large number of features (vocabulary size) very efficiently and quickly.

- The independence assumption, while flawed, often provides acceptable performance for text-based tasks.

19. Compare SVM and Naïve Bayes for classification tasks.

- SVM: A discriminative model that focuses on finding the best separating hyperplane to maximize the margin. It is generally highly accurate but computationally expensive and sensitive to feature scaling and kernel choice.

- Naïve Bayes: A generative model based on probability and the independence assumption. It is much faster, scales better with data size, and is excellent for high-dimensional, sparse data like text, but its accuracy can be limited by its rigid assumptions.

20. How does Laplace Smoothing help in Naïve Bayes?
- Laplace Smoothing (or additive smoothing) is a technique used to address the Zero-Frequency problem. If a feature value (e.g., a word) is present in the test set but not in the training set, its conditional probability becomes zero, which incorrectly makes the entire posterior probability zero. Smoothing prevents this by adding a small positive value (usually 1) to all counts, ensuring that all features have a non-zero probability.

```python
# 21. Write a Python program to train an SVM Classifier on the Iris dataset and evaluate accuracy:
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load Data
X, y = load_iris(return_X_y=True)

# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. Train model
# Using linear kernel as a starting point
svm_classifier = SVC(kernel = 'linear', random_state = 42)
svm_classifier.fit(X_train, y_train)

# 4. Predict and Evaluate
y_pred = svm_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"SVM Accuracy: {accuracy:.4f}")
```

```python
# 22. Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies.

import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# 1. Load Data
X, y = load_wine(return_X_y=True)

# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. Scale Features (Crucial for RBF kernel to prevent features with large values from dominating)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# --- Train and Evaluate with Linear Kernel ---
# Linear kernel assumes data is linearly separable.
svc_linear = SVC(kernel='linear', random_state=42)
svc_linear.fit(X_train_scaled, y_train)

y_pred_linear = svc_linear.predict(X_test_scaled)
accuracy_linear = accuracy_score(y_test, y_pred_linear)


# --- Train and Evaluate with RBF Kernel ---
# RBF kernel handles non-linear relationships by mapping data to a higher dimension.
# We set C and gamma for better performance.
svc_rbf = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)
svc_rbf.fit(X_train_scaled, y_train)

y_pred_rbf = svc_rbf.predict(X_test_scaled)
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)


# 4. Compare Accuracies
print("--- SVM Kernel Comparison (Wine Dataset) ---")
print(f"Accuracy with Linear Kernel: {accuracy_linear:.4f}")
print(f"Accuracy with RBF Kernel: {accuracy_rbf:.4f}")

if accuracy_rbf > accuracy_linear:
    print("\nConclusion: RBF kernel performed better, suggesting a non-linear boundary.")
else:
    print("\nConclusion: Linear kernel performed better or equally, suggesting a close-to-linear boundary.")
```

```python
# 24. Write a Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision boundary.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# --- 1. Load and Prepare Data ---
# Create a non-linearly separable dataset (concentric circles)
X, y = make_circles(n_samples=100, factor=0.5, noise=0.1, random_state=42)

# Split data (using 70% for training)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Scale features (important for SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


# --- 2. Train Model with Polynomial Kernel ---
# Use a polynomial kernel of degree 3
poly_svc = SVC(kernel='poly', degree=3, gamma='auto', C=1.0, random_state=42)
poly_svc.fit(X_train_scaled, y_train)


# --- 3. Visualization Function ---

def plot_decision_boundary(model, X, y, title):
    # Define the range for the plot (based on the entire scaled data)
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

    # Create a meshgrid to evaluate the model across the entire feature space
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))

    # Predict the class for every point in the meshgrid
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Plot the contour and the training points
    plt.figure(figsize=(8, 6))
    plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.coolwarm)

    # Plot the training points (support vectors will be highlighted later if needed)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm)

    plt.title(title)
    plt.xlabel('Feature 1 (Scaled)')
    plt.ylabel('Feature 2 (Scaled)')
    plt.show()

# --- 4. Visualize the Result ---
plot_decision_boundary(
    poly_svc,
    np.vstack((X_train_scaled, X_test_scaled)), # Use all scaled data for boundary limits
    np.hstack((y_train, y_test)),
    'SVC with Polynomial Kernel (Degree 3) Decision Boundary'
)
```

```python
# 26. Write a Python program to train a Multinomial Naïve Bayes classifier for text classification using the 20 Newsgroups Dataset.

from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Load Data (Subset of categories for simplicity)
categories = ['alt.atheism', 'soc.religion.christian', 'comp.graphics', 'sci.med']
newsgroups_train = fetch_20newsgroups(subset = 'train', categories = categories, shuffle = True, random_state = 42)
newsgroups_test = fetch_20newsgroups(subset = 'test', categories = categories, shuffle = True, random_state =42)

# 2. Feature Extraction (Convert text to Tf-IDF vectors)
vectorizer = TfidfVectorizer()
# Fit the vectorizer on the training data and transform both training and testing data
X_train = vectorizer.fit_transform(newsgroups_train.data)
X_test = vectorizer.transform(newsgroups_test.data) # Use transform here
y_train = newsgroups_train.target
y_test = newsgroups_test.target

# 3. Train Model
# MultiniomialNB is ideal for text data(count/frequency features).
mnb_classifier = MultinomialNB()
mnb_classifier.fit(X_train, y_train)

# 4. Predict and Evaluate
y_pred = mnb_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Multinomial Naive Bayes Accuracy: {accuracy:.4f}")
```

```python
# 27. Write a Python program to train an SVM Classifier with different C values and compare the decision boundaries visually.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load and Prepare Data
# Create a non-linear dataset with some intermingling (simulating outliers/noise)
X, y = make_moons(n_samples=100, noise=0.2, random_state=42)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define C values to test: Low, Medium, High
C_values = [0.1, 1, 100]


# 2. Visualization Setup
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
x_min, x_max = X_scaled[:, 0].min() - 0.1, X_scaled[:, 0].max() + 0.1
y_min, y_max = X_scaled[:, 1].min() - 0.1, X_scaled[:, 1].max() + 0.1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

# 3. Train Models and Plot
for C, ax in zip(C_values, axes):
    # Train RBF kernel SVC with different C values
    model = SVC(kernel='rbf', C=C, gamma='auto', random_state=42)
    model.fit(X_scaled, y)

    # Predict over the meshgrid
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Plot results
    ax.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.coolwarm)
    # Plot data points, highlighting support vectors (which define the boundary)
    ax.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolors='k')

    # Highlight Support Vectors
    svs = model.support_vectors_
    ax.scatter(svs[:, 0], svs[:, 1], s=180, facecolors='none', edgecolors='k', marker='o')

    # Define title inside the loop to use the current C value
    title = f'C = {C} (Large Margin, More Slack)' if C == 0.1 else (f'C = {C} (Default Balance)' if C == 1 else f'C = {C} (Small Margin, Less Slack)')
    ax.set_title(title)
    ax.set_xticks(())
    ax.set_yticks(())

plt.suptitle("Effect of C Parameter on SVM Decision Boundary", fontsize=16)
plt.show()
```

```python
# 29. Write a Python program to apply feature scaling before training an SVM model and compare results with unscaled data.

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
import warnings

# Suppress ConvergenceWarning from SVC on unscaled data
warnings.filterwarnings("ignore", category=UserWarning)

# 1. Load Data
X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# --- Scenario 1: Unscaled Data ---
svc_unscaled = SVC(kernel='rbf', gamma='auto', random_state=42)
svc_unscaled.fit(X_train, y_train)
acc_unscaled = accuracy_score(y_test, svc_unscaled.predict(X_test))

# --- Scenario 2: Scaled Data ---
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svc_scaled = SVC(kernel='rbf', gamma='auto', random_state=42)
svc_scaled.fit(X_train_scaled, y_train)
acc_scaled = accuracy_score(y_test, svc_scaled.predict(X_test_scaled))

# 3. Comparison
print("--- Effect of Feature Scaling on RBF-Kernel SVM ---")
print(f"Accuracy with Unscaled Data: {acc_unscaled:.4f}")
print(f"Accuracy with Scaled Data (StandardScaler): {acc_scaled:.4f}")

if acc_scaled > acc_unscaled:
    print("\nConclusion: Feature scaling significantly improved SVM accuracy.")
else:
    print("\nConclusion: Feature scaling made little difference (often due to highly linear data or perfect scaling by chance")
```

```python
# 30. Write a Python program to train a Gaussian Naïve Bayes model and compare the predictions before and after Laplace Smoothing?

import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_iris

# 1. Load Data (Iris data has continuous features suitable for GaussianNB)
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# --- Scenario 1: Default Smoothing (var_smoothing is 1e-9) ---
# This is usually sufficient to prevent zero-variance issues.
gnb_default = GaussianNB()
gnb_default.fit(X_train, y_train)
y_pred_default = gnb_default.predict(X_test)
acc_default = accuracy_score(y_test, y_pred_default)

# --- Scenario 2: Increased Smoothing (Over-smoothing) ---
# A large var_smoothing value increases the assumed variance,
# making the model more robust to outliers but potentially increasing bias.
gnb_smooth = GaussianNB(var_smoothing=1.0) # Significantly higher smoothing
gnb_smooth.fit(X_train, y_train)
y_pred_smooth = gnb_smooth.predict(X_test)
acc_smooth = accuracy_score(y_test, y_pred_smooth)

# 3. Comparison of Predictions
print("--- Gaussian Naïve Bayes Smoothing Comparison ---")
print(f"Accuracy (Default Smoothing): {acc_default:.4f}")
print(f"Accuracy (High Smoothing, var_smoothing=1.0): {acc_smooth:.4f}")

# Example of comparing raw predictions
print("\nFirst 5 Predictions (Default vs. Smooth):")
print(f"True Labels:   {y_test[:5]}")
print(f"Default Preds: {y_pred_default[:5]}")
print(f"Smooth Preds:  {y_pred_smooth[:5]}")

if acc_smooth != acc_default:
    print("\nConclusion: The smoothing parameter had a measurable effect on the predictions.")
else:
    print("\nConclusion: The smoothing parameter had minimal effect on the final accuracy in this case.")
```

```python
# 32. Write a Python program to train an SVM Classifier on an imbalanced dataset and apply class weighting and check it improve accuracy?

import numpy as np
from sklearn.svm import SVC
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score

# 1. Create Imbalanced Data
X, y = make_classification(n_samples = 1000, n_features = 10, n_informative = 5, n_redundant = 5,n_classes = 2, n_clusters_per_class =1, weights = [0.95,0.05], flip_y = 0, random_state = 42)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 2. Train Model 1: Without Class Weighting
svm_unweighted = SVC(kernel = 'linear', random_state = 42)
svm_unweighted.fit(X_train, y_train)
y_pred_unweighted = svm_unweighted.predict(X_test)

# 3. Train Model 3: With Class Weighting
svm_weighted = SVC(kernel = 'linear', class_weight = 'balanced',  random_state = 42)
svm_weighted.fit(X_train, y_train)
y_pred_weighted = svm_weighted.predict(X_test)

# 4. Compare Performance (Accuracy is misleading; F1-Score is better for imbalance)
f1_unweighted = f1_score(y_test, y_pred_unweighted, average='binary')
f1_weighted = f1_score(y_test, y_pred_weighted, average='binary')
acc_unweighted = accuracy_score(y_test, y_pred_unweighted)
acc_weighted = accuracy_score(y_test, y_pred_weighted)

print("--- SVM on Imbalanced Data with Class Weighting ---")
print(f"Baseline Accuracy (Unweighted): {acc_unweighted:.4f}")
print(f"Weighted Accuracy: {acc_weighted:.4f}")
print("-" * 35)
print(f"Baseline F1-Score (Unweighted): {f1_unweighted:.4f}")
print(f"Weighted F1-Score: {f1_weighted:.4f}")

if f1_weighted > f1_unweighted:
    print("\nConclusion: Class weighting improved the F1-Score, indicating better minority class prediction.")
else:
    print("\nConclusion: Class weighting did not improve F1-Score in this test case.")
```

```python
# 36. Write a Python program to train an SVM Classifier using One-vs-Rest (OvR) and One-vs-One (OvO) strategies on the Wine dataset and compare their accuracy.


from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

# 1. Load Data
X, y = load_wine(return_X_y = True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 2. Scale Data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Strategy1: One-Vs-Rest
# Trains K (3) binary classifiers
ovr_model = OneVsRestClassifier(SVC(kernel = 'linear', random_state = 42))
ovr_model.fit(X_train_scaled, y_train)
y_pred_ovr = ovr_model.predict(X_test_scaled)
acc_ovr = accuracy_score(y_test, y_pred_ovr)

# Strategy2: One-Vs-One
# Trains K*(k-1)/2 (3) binary classifiers
ovo_model = OneVsOneClassifier(SVC(kernel = 'linear', random_state = 42))
ovo_model.fit(X_train, y_train)
y_pred_ovo = ovo_model.predict(X_test)
acc_ovo = accuracy_score(y_test, y_pred_ovo)

# 3. Compare Results
print("--- SVM Multiclass Strategy Comparison (Wine Dataset) ---")
print(f"Accuracy with One-vs-Rest (OvR): {acc_ovr:.4f}")
print(f"Accuracy with One-vs-One (OvO): {acc_ovo:.4f}")

if acc_ovo == acc_ovr:
    print("\nConclusion: Both OvR and OvO strategies yielded the same accuracy for this dataset.")
else:
    print("\nConclusion: OvO and OvR yielded different accuracies.")
```

```python
# 40.Write a Python program to perform Recursive Feature Elimination (RFE) before training an SVM Classifier and compare accuracy.

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import RFE
from sklearn.preprocessing import StandardScaler

# 1. Load Data
X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Scale data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
n_features_total = X.shape[1]
n_features_to_select = 5 # Target number of features

# --- Scenario 1: SVM on All Features ---
svm_all = SVC(kernel='linear', random_state=42)
svm_all.fit(X_train_scaled, y_train)
acc_all = accuracy_score(y_test, svm_all.predict(X_test_scaled))


# --- Scenario 2: SVM on RFE-Selected Features ---
# Use a linear SVC inside RFE for feature ranking
rfe_estimator = SVC(kernel='linear')
rfe_selector = RFE(rfe_estimator, n_features_to_select=n_features_to_select, step=1)
rfe_selector.fit(X_train_scaled, y_train)

# Transform data to include only the selected features
X_train_rfe = rfe_selector.transform(X_train_scaled)
X_test_rfe = rfe_selector.transform(X_test_scaled)

# Train the final SVM model on the reduced features
svm_rfe = SVC(kernel='linear', random_state=42)
svm_rfe.fit(X_train_rfe, y_train)
acc_rfe = accuracy_score(y_test, svm_rfe.predict(X_test_rfe))

# 3. Compare Results
print("--- Recursive Feature Elimination (RFE) for SVM ---")
print(f"Total features: {n_features_total}")
print(f"Features selected by RFE: {n_features_to_select}")
print("-" * 35)
print(f"Accuracy with All Features: {acc_all:.4f}")
print(f"Accuracy with RFE-Selected Features: {acc_rfe:.4f}")

if acc_rfe > acc_all:
    print("\nConclusion: RFE improved accuracy by removing uninformative features.")
elif acc_rfe < acc_all:
    print("\nConclusion: RFE slightly hurt accuracy but reduced model complexity.")
else:
    print("\nConclusion: RFE provided similar accuracy with fewer features.")
```
